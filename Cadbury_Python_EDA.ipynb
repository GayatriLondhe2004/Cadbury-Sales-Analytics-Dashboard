```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("Cadbury_Business_Style_Dataset.xlsx")

df_clean = df.copy()

df_clean["Date"] = pd.to_datetime(df_clean["Date"])

PURPLE = "#6B21A8"
TEAL = "#0F6E56"

COLORS = ["#6B21A8","#1D9E75","#D85A30","#378ADD","#BA7517",
          "#A32D2D","#3B6D11","#854F0B","#993556","#5F5E5A"]
```

```python
plt.figure(figsize=(10,6))

rev_prod = (
    df_clean.groupby("Product")["Revenue"]
    .sum()
    .sort_values()
    / 1000000
)

plt.barh(rev_prod.index, rev_prod.values, color=COLORS)

plt.title("Total Revenue by Product (₹ M)")
plt.xlabel("Revenue (₹ Millions)")
plt.ylabel("Product")

plt.show()
```

```python
plt.figure(figsize=(7,7))

rev_region = df_clean.groupby("Region")["Revenue"].sum()

plt.pie(
    rev_region,
    labels=rev_region.index,
    autopct="%1.1f%%",
    startangle=140
)

plt.title("Revenue Distribution by Region")

plt.show()
```

```python
month_order = [
    "Jan","Feb","Mar","Apr","May","Jun",
    "Jul","Aug","Sep","Oct","Nov","Dec"
]

rev_month = (
    df_clean.groupby("Month")["Revenue"]
    .sum()
    .reindex(month_order)
    / 1000000
)

plt.figure(figsize=(10,5))

plt.plot(
    rev_month.index,
    rev_month.values,
    marker="o"
)

plt.fill_between(
    rev_month.index,
    rev_month.values,
    alpha=0.3
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹ Millions)")

plt.show()
```

```python
rev_channel = (
    df_clean.groupby("Channel")["Revenue"]
    .sum()
    / 1000000
)

plt.figure(figsize=(8,5))

plt.bar(
    rev_channel.index,
    rev_channel.values
)

plt.title("Revenue by Sales Channel")
plt.ylabel("Revenue (₹ Millions)")

plt.show()
```
